# Probability Distributions in Python

This notebook creates, visualizes, and samples from several common probability distributions.

The discrete distributions are:

- Discrete uniform
- Bernoulli
- Binomial
- Geometric
- Poisson

The continuous distributions are:

- Gaussian (normal)
- Exponential

For a discrete random variable, the plots show the **probability mass function (PMF)**. For a continuous random variable, they show the **probability density function (PDF)**. The final section compares simulated sample statistics with theoretical means and variances.

## Setup

This notebook uses only the Python standard library, NumPy, and Matplotlib.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams["figure.figsize"] = (8, 4)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
```

## Plotting Helpers

PMFs are displayed with bars because probability is assigned to individual outcomes. PDFs are displayed as curves because probabilities are areas over intervals.

```python
def plot_pmf(x_values, probabilities, title, xlabel, color="#4c78a8"):
    """Plot a probability mass function as a bar chart."""
    fig, ax = plt.subplots()
    ax.bar(x_values, probabilities, color=color, edgecolor="black", alpha=0.85)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Probability")
    ax.set_ylim(0, max(probabilities) * 1.2)
    ax.set_xticks(x_values)
    plt.show()


def plot_pdf(x_values, densities, title, xlabel, color="#4c78a8"):
    """Plot a probability density function as a curve."""
    fig, ax = plt.subplots()
    ax.plot(x_values, densities, color=color, linewidth=2.5)
    ax.fill_between(x_values, densities, color=color, alpha=0.2)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Probability density")
    ax.set_ylim(bottom=0)
    plt.show()
```

## 1. Discrete Uniform Distribution

A discrete uniform distribution assigns equal probability to every outcome in a finite set. A fair six-sided die is an example:

$$
P(X=k)=\frac{1}{6},
\qquad k\in\{1,2,3,4,5,6\}.
$$

```python
die_faces = np.arange(1, 7)
uniform_probs = np.full(len(die_faces), 1 / len(die_faces))

plot_pmf(
    die_faces,
    uniform_probs,
    title="Discrete Uniform Distribution: Fair Six-Sided Die",
    xlabel="Die outcome",
    color="#4c78a8",
)
```

## 2. Bernoulli Distribution

A Bernoulli distribution models one trial with two possible outcomes: failure ($0$) and success ($1$). Its parameter $p$ is the probability of success:

$$
P(X=1)=p,
\qquad
P(X=0)=1-p.
$$

```python
bernoulli_p = 0.65
bernoulli_outcomes = np.array([0, 1])
bernoulli_probs = np.array([1 - bernoulli_p, bernoulli_p])

plot_pmf(
    bernoulli_outcomes,
    bernoulli_probs,
    title=f"Bernoulli Distribution (p = {bernoulli_p})",
    xlabel="Outcome",
    color="#f58518",
)
```

## 3. Binomial Distribution

A Binomial distribution models the number of successes in $n$ independent Bernoulli trials with constant success probability $p$:

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}.
$$

```python
binomial_n = 10
binomial_p = 0.4
success_counts = np.arange(0, binomial_n + 1)
binomial_probs = np.array([
    math.comb(binomial_n, k)
    * (binomial_p ** k)
    * ((1 - binomial_p) ** (binomial_n - k))
    for k in success_counts
])

plot_pmf(
    success_counts,
    binomial_probs,
    title=f"Binomial Distribution (n = {binomial_n}, p = {binomial_p})",
    xlabel="Number of successes",
    color="#54a24b",
)
```

## 4. Geometric Distribution

A Geometric distribution models the number of independent Bernoulli trials needed to obtain the first success:

$$
P(X=k)=(1-p)^{k-1}p,
\qquad k=1,2,\ldots.
$$

The plot is truncated at 15 trials, although the distribution has an infinite right tail.

```python
geometric_p = 0.3
trial_numbers = np.arange(1, 16)
geometric_probs = np.array([
    ((1 - geometric_p) ** (k - 1)) * geometric_p
    for k in trial_numbers
])

plot_pmf(
    trial_numbers,
    geometric_probs,
    title=f"Geometric Distribution (p = {geometric_p})",
    xlabel="Trial number of first success",
    color="#e45756",
)
```

## 5. Poisson Distribution

A Poisson distribution models the number of events in a fixed interval when events occur independently at a constant average rate $\lambda$:

$$
P(X=k)=e^{-\lambda}\frac{\lambda^k}{k!},
\qquad k=0,1,2,\ldots.
$$

```python
poisson_rate = 4
event_counts = np.arange(0, 16)
poisson_probs = np.array([
    (math.exp(-poisson_rate) * (poisson_rate ** k)) / math.factorial(k)
    for k in event_counts
])

plot_pmf(
    event_counts,
    poisson_probs,
    title=f"Poisson Distribution (lambda = {poisson_rate})",
    xlabel="Number of events",
    color="#b279a2",
)
```

## 6. Gaussian Distribution

A Gaussian distribution models a continuous quantity concentrated around a mean $\mu$. Its standard deviation $\sigma$ controls its spread:

$$
p(x)=\frac{1}{\sigma\sqrt{2\pi}}
\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right).
$$

```python
gaussian_mu = 0.0
gaussian_sigma = 1.5
gaussian_x = np.linspace(
    gaussian_mu - 4 * gaussian_sigma,
    gaussian_mu + 4 * gaussian_sigma,
    500,
)
gaussian_pdf = (
    np.exp(-0.5 * ((gaussian_x - gaussian_mu) / gaussian_sigma) ** 2)
    / (gaussian_sigma * np.sqrt(2 * np.pi))
)

plot_pdf(
    gaussian_x,
    gaussian_pdf,
    title=f"Gaussian Distribution (mu = {gaussian_mu}, sigma = {gaussian_sigma})",
    xlabel="x",
    color="#72b7b2",
)
```

The mean and variance are

$$
\mathbb{E}[X]=\mu,
\qquad
\operatorname{Var}(X)=\sigma^2.
$$

## 7. Exponential Distribution

An Exponential distribution models a nonnegative waiting time. Its rate parameter $\lambda$ controls how quickly the density decreases:

$$
p(x)=\lambda e^{-\lambda x},
\qquad x\geq0.
$$

Its CDF is $F(x)=1-e^{-\lambda x}$. The following code plots both the PDF and CDF.

```python
exponential_rate = 0.5
exponential_x = np.linspace(0, 12, 500)
exponential_pdf = exponential_rate * np.exp(-exponential_rate * exponential_x)
exponential_cdf = 1 - np.exp(-exponential_rate * exponential_x)

fig, ax = plt.subplots()
ax.plot(
    exponential_x,
    exponential_pdf,
    label="PDF",
    color="#e45756",
    linewidth=2.5,
)
ax.plot(
    exponential_x,
    exponential_cdf,
    label="CDF",
    color="#4c78a8",
    linewidth=2.5,
)
ax.set_title(f"Exponential Distribution (lambda = {exponential_rate})")
ax.set_xlabel("Waiting time")
ax.set_ylabel("Density or cumulative probability")
ax.set_ylim(0, 1.05)
ax.legend()
plt.show()
```

The mean and variance are

$$
\mathbb{E}[X]=\frac{1}{\lambda},
\qquad
\operatorname{Var}(X)=\frac{1}{\lambda^2}.
$$

## Generate Random Samples

The previous cells plot theoretical PMFs and PDFs. NumPy can also generate random samples from each distribution. With a sufficiently large sample, the sample mean and variance should be close to their theoretical values.

```python
rng = np.random.default_rng(seed=42)
sample_size = 10_000

samples = {
    "uniform_die": rng.integers(1, 7, size=sample_size),
    "bernoulli": rng.binomial(1, bernoulli_p, size=sample_size),
    "binomial": rng.binomial(binomial_n, binomial_p, size=sample_size),
    "geometric": rng.geometric(geometric_p, size=sample_size),
    "poisson": rng.poisson(poisson_rate, size=sample_size),
    "gaussian": rng.normal(gaussian_mu, gaussian_sigma, size=sample_size),
    "exponential": rng.exponential(1 / exponential_rate, size=sample_size),
}

theoretical_values = {
    "uniform_die": {
        "mean": (1 + 6) / 2,
        "variance": (6**2 - 1) / 12,
    },
    "bernoulli": {
        "mean": bernoulli_p,
        "variance": bernoulli_p * (1 - bernoulli_p),
    },
    "binomial": {
        "mean": binomial_n * binomial_p,
        "variance": binomial_n * binomial_p * (1 - binomial_p),
    },
    "geometric": {
        "mean": 1 / geometric_p,
        "variance": (1 - geometric_p) / (geometric_p**2),
    },
    "poisson": {
        "mean": poisson_rate,
        "variance": poisson_rate,
    },
    "gaussian": {
        "mean": gaussian_mu,
        "variance": gaussian_sigma**2,
    },
    "exponential": {
        "mean": 1 / exponential_rate,
        "variance": 1 / (exponential_rate**2),
    },
}

print(
    f"{'Distribution':12s} "
    f"{'Sample Mean':>12s} "
    f"{'Theory Mean':>12s} "
    f"{'Sample Var':>12s} "
    f"{'Theory Var':>12s}"
)
print("-" * 66)

for name, values in samples.items():
    theory = theoretical_values[name]
    print(
        f"{name:12s} "
        f"{values.mean():12.3f} "
        f"{theory['mean']:12.3f} "
        f"{values.var():12.3f} "
        f"{theory['variance']:12.3f}"
    )
```

## Compare Continuous Samples with Their PDFs

The histograms below estimate the distributions from simulated data. The theoretical PDFs are overlaid as curves.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(
    samples["gaussian"],
    bins=40,
    density=True,
    alpha=0.55,
    color="#72b7b2",
    edgecolor="white",
    label="Samples",
)
axes[0].plot(
    gaussian_x,
    gaussian_pdf,
    color="black",
    linewidth=2,
    label="Theoretical PDF",
)
axes[0].set_title("Gaussian Samples and PDF")
axes[0].set_xlabel("x")
axes[0].set_ylabel("Density")
axes[0].legend()

axes[1].hist(
    samples["exponential"],
    bins=40,
    density=True,
    alpha=0.55,
    color="#e45756",
    edgecolor="white",
    label="Samples",
)
axes[1].plot(
    exponential_x,
    exponential_pdf,
    color="black",
    linewidth=2,
    label="Theoretical PDF",
)
axes[1].set_xlim(0, 12)
axes[1].set_title("Exponential Samples and PDF")
axes[1].set_xlabel("Waiting time")
axes[1].set_ylabel("Density")
axes[1].legend()

plt.tight_layout()
plt.show()
```
