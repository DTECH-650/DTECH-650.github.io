## Variables and built-in types

A variable associates a name with a value. Python determines the type at runtime. Common built-in types include integers, floating-point numbers, complex numbers, strings, Boolean values, and the special value None.

Choose descriptive names, use lowercase words separated by underscores, and remember that names are case-sensitive.

````python
# --- INTEGER DEFINITION ---
# If no decimal place is provided, Python considers it an integer
a = 9

# --- FLOAT DEFINITION ---
# Specifying a decimal point creates a float
b = 6.05

# --- STRING DEFINITION ---
# Strings are collections of characters, defined with '' or ""
c = 'for'
d = "good"

# --- BOOL DEFINITION ---
# Booleans represent True or False values
e = True

# --- NoneType DEFINITION ---
# None represents the absence of a value
f = None

# You can print this with the type() function
print("Datatype of variable a = ", type(a))
print("Datatype of variable b = ", type(b))
print("Datatype of variable c = ", type(c))
print("Datatype of variable d = ", type(d))
print("Datatype of variable e = ", type(e))
print("Datatype of variable f = ", type(f))
````

````python
course_name = "TECH 65000"    # str
week = 1                     # int (if you assign a whole number without decimal points python automatically treats it as an integer)
temperature = 22.5           # float (if you assign a number with a decimal point python automatically treats it as a float)
is_ready = True              # bool
missing_value = None         # NoneType

# a for loop to iterate over all the variables and print their values and types (We turn to for loops for iteration later in the tutorial when we discuss control flow)
for value in (course_name, week, temperature, is_ready, missing_value):
    print(f"{value!r:12} -> {type(value).__name__}") 
# f-string formatting: !r uses repr(value), :12 sets a minimum field width,
# and __name__ returns the type name.
# See: https://docs.python.org/3/tutorial/inputoutput.html#formatted-string-literals
````

## Collections (Python built-in)

Python's main built-in collections are:

- **list:** ordered and mutable;
- **tuple:** ordered and immutable;
- **dictionary:** mutable key-value pairs; and
- **set:** an unordered collection of unique values.

Indices begin at zero. A slice start:stop includes the start position but excludes the stop position.

--> We return to numpy arrays later in this tutorial! 

````python
measurements = [2.4, 2.7, 2.5, 2.8] #an example of a list
point = (3.0, 4.0) #an example of a tuple
student = {"name": "Ada", "score": 92} #an example of a dictionary
labels = {"train", "test", "train"} #an example of a set

# --- PRINTING INITIAL COLLECTIONS (and specific entries in it)---
print("Initial collections:")
print("first two measurements:", measurements[:2]) # prints the first two elements of the list
print("point:", point) # prints the entire tuple
print("student:", student) # prints the entire dictionary
print("unique labels:", labels) # prints the entire set
print("------------------------------")

# --- MODIFYING COLLECTIONS ---
# You can add elements to lists and dictionaries as shown below

# Adding an element to the list
measurements.append(3.0)

# Adding a key-value pair to the dictionary
student["passed"] = student["score"] >= 70

# --- PRINTING COLLECTIONS AFTER ADDITIONS ---
print("Collections after additions:")
print("measurements:", measurements) # prints the entire list
print("point:", point) # prints the entire tuple
print("student:", student) # prints the entire dictionary
print("unique labels:", labels) # prints the entire set
print("------------------------------")


# You can also remove elements from lists and dictionaries
# Removing an element from the list
measurements.pop()  # removes the last element

# Removing a key-value pair from the dictionary
del student["passed"]
# --- PRINTING COLLECTIONS AFTER DELETIONS ---
print("Collections after deletions:")
print("measurements:", measurements) # prints the entire list
print("point:", point) # prints the entire tuple
print("student:", student) # prints the entire dictionary
print("unique labels:", labels) # prints the entire set
print("------------------------------")
````

## When to use what? 
“Do I need an ordered sequence that I can modify?” → list

“Do I need an ordered group that should not change?” → tuple

“Do I want to look things up by a meaningful key?” → dict

“Do I care only whether a unique value is present?” → set

# Control Flow statements
Until now, Python executed our statements sequentially from top to bottom. Control-flow statements let the program decide which statements to execute and how often to execute them.

## Conditional statements

An if statement selects which block of code runs. Indentation is part of Python's syntax, so use four spaces for each level.

````python
score = 84

if score >= 90:
    letter = "A"
elif score >= 80:
    letter = "B"
elif score >= 70:
    letter = "C"
else:
    letter = "Needs improvement"

print(letter)
````

## Loops

A for loop processes each item in an iterable. A while loop repeats as long as its condition remains true. The built-ins range and enumerate are especially useful in loops.

````python
values = [3, 5, 8]

print("Initial values:", values)

print("------------------------------")

print("A for loop to iterate over the values and print their squares:")

for index, value in enumerate(values):
    print(f"index {index}: square = {value ** 2}")
print("------------------------------")

print("A while loop to count down from 3 to 1:")
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Countdown finished.")
print("------------------------------")
````

## Functions

A function packages reusable behavior. Parameters are inputs, return sends a result back to the caller, and a docstring explains the function. Type hints document expected types but do not automatically enforce them.

````python

print("Demonstrating functions in Python")
print("------------------------------")

# Define a function to calculate the mean of a list of numbers and a function to greet a person with an optional greeting message.
def mean(values: list[float]) -> float:
    """Return the arithmetic mean of a nonempty list."""
    return sum(values) / len(values)


def greet(name: str, greeting: str = "Hello") -> str:
    return f"{greeting}, {name}!"


#### Example usage of the functions defined above

print("Without functions, we repeat the calculation:")

values_1 = [2.0, 4.0, 9.0]
result_1 = sum(values_1) / len(values_1)
print("Mean of first list:", result_1)

values_2 = [5.0, 7.0, 10.0, 12.0]
result_2 = sum(values_2) / len(values_2)
print("Mean of second list:", result_2)

print("------------------------------")

print("Now we use a function so the calculation is defined only once.")

print("Mean of first list:", mean(values_1))
print("Mean of second list:", mean(values_2))

print("------------------------------")

print("Functions can also accept different inputs:")

print(greet("Grace"))                     # default greeting
print(greet("Linus", greeting="Welcome")) # custom greeting
print(greet("Ada", greeting="Good morning"))
````

### Scope

Names created inside a function are normally local to that function. Prefer returning results over changing global variables; this makes code easier to test and reuse.

````python
scale = 10

def scaled_square(value: float) -> float:
    local_result = scale * value ** 2
    
    print("Inside function:")
    print("value =", value)
    print("scale =", scale)
    print("local_result =", local_result)
    
    return local_result

result = scaled_square(3)

print("Outside function:")
print("scale =", scale)
print("result =", result)

#print(local_result)  # This will raise an error because local_result is not defined outside the function.
````

## Comprehensions

A comprehension constructs a collection from an iterable in a compact form. Use one when the transformation remains easy to read; otherwise, use a regular loop.

````python
squares = [number ** 2 for number in range(6)] # list comprehension to create a list of squares
even_squares = [value for value in squares if value % 2 == 0] # list comprehension to filter even squares
square_lookup = {number: number ** 2 for number in range(4)} # dictionary comprehension to map numbers to their squares

print("Demonstrating list and dictionary comprehensions in Python")
print("------------------------------")
print(f"created list of squares: {squares}")
print(f"filtered list of even squares: {even_squares}")
print(f"dictionary mapping numbers to their squares: {square_lookup}")
````

````python
# Equivalent versions using regular loops

# 1. Create a list of squares
squares_loop = []

for number in range(6):
    squares_loop.append(number ** 2)


# 2. Keep only the even squares
even_squares_loop = []

for value in squares_loop:
    if value % 2 == 0:
        even_squares_loop.append(value)


# 3. Create a dictionary mapping each number to its square
square_lookup_loop = {}

for number in range(4):
    square_lookup_loop[number] = number ** 2


print(squares_loop)
print(even_squares_loop)
print(square_lookup_loop)
````

## Exceptions

Python raises an exception when it cannot complete an operation. Read the final line of a traceback first: it identifies the exception type and usually explains the problem. Catch only exceptions you can handle meaningfully.

````python
raw_value = "not a number"

try:
    number = float(raw_value) # Attempt to convert the raw value to a float
except ValueError:
    print(f"Error: {raw_value!r} cannot be converted to a float.") # Handle the case where conversion fails
else:
    print(number) # Print the successfully converted number
finally:
    print("Conversion attempt complete.") # This block always executes
````

## Classes

A class defines a new type that combines data with behavior. An object is an instance of a class. The constructor initializes each object, and methods operate on its data.

````python
class Vector2D:
    """Represent a vector in two-dimensional space."""

    def __init__(self, x: float, y: float) -> None:
        """
        Initialize a new Vector2D object.

        Parameters
        ----------
        x : float
            Horizontal component of the vector.
        y : float
            Vertical component of the vector.
        """
        self.x = x
        self.y = y

    def norm(self) -> float:
        """
        Return the Euclidean magnitude of the vector.

        The magnitude is

            sqrt(x^2 + y^2)
        """
        return (self.x ** 2 + self.y ** 2) ** 0.5


# ------------------------------------------------------------
# Create two different instances of the same class
# ------------------------------------------------------------

print("Creating two Vector2D instances:")
vector_1 = Vector2D(3.0, 4.0)
vector_2 = Vector2D(1.0, 2.0)
print("-------------------------")


# ------------------------------------------------------------
# Access attributes stored in each object
# ------------------------------------------------------------
print("Accessing attributes of the two Vector2D instances:")
print("Vector 1")
print("x-coordinate:", vector_1.x)
print("y-coordinate:", vector_1.y)

print()
print("-------------------------")

print("Vector 2")
print("x-coordinate:", vector_2.x)
print("y-coordinate:", vector_2.y)
print("-------------------------")


# ------------------------------------------------------------
# Call a method defined by the class
# ------------------------------------------------------------

print("Calling methods of the two Vector2D instances:")
print("Norm of vector 1:", vector_1.norm()) # this is similar to Vector2D.norm(vector_1)
print("Norm of vector 2:", vector_2.norm())
print("-------------------------")


# ------------------------------------------------------------
# Inspect the type of the objects
# ------------------------------------------------------------

print("Inspecting the type of the two Vector2D instances:")
print("Type of vector_1:", type(vector_1))
print("Is vector_1 a Vector2D object?", isinstance(vector_1, Vector2D))
print("Type of vector_2:", type(vector_2))
print("Is vector_2 a Vector2D object?", isinstance(vector_2, Vector2D))
print("-------------------------")

# The above code demonstrates creating a simple 2D vector class, instantiating it, and using its methods.
# Python classes tutorial:
# https://docs.python.org/3/tutorial/classes.html
````

## NumPy arrays
NumPy is a Python library for numerical computing. Its central data structure is the NumPy array (ndarray).

### Why NumPy? 
A NumPy array is similar to a Python list in that it stores multiple values, but it is designed specifically for numerical computation. NumPy arrays usually contain values of a common data type and support efficient mathematical operations on entire collections of values.

This becomes especially important in statistics and machine learning, where data are commonly represented as vectors, matrices, and higher-dimensional arrays.

For example:

- a single observation may be represented as a vector;
- an entire dataset may be represented as a matrix;
- model parameters may be stored as vectors or matrices; and
- numerical operations are often applied to many values simultaneously.

NumPy provides efficient multidimensional arrays and numerical operations. Unlike a list, an array usually stores values of one data type. 

## Key NumPy Operations
Operations can act element by element without an explicit Python loop.

For example: 
> The @ operator performs matrix multiplication; 
> * performs element-wise multiplication.

````python
import numpy as np

# A regular Python list
python_values = [1, 2, 3]

# A NumPy array containing the same values
numpy_values = np.array([1, 2, 3])

print("Demonstrating the difference between Python lists and NumPy arrays")

print("Python list:", python_values)
print("NumPy array:", numpy_values)

print("Python type:", type(python_values))
print("NumPy type:", type(numpy_values))

# Why is Numpy useful? It performs element-wise operations efficiently, unlike regular Python lists.

print("--------------------------------")
print("Demonstrating element-wise operations")

print("Python list * 2:")
print(python_values * 2)

print()

print("NumPy array * 2:")
print(numpy_values * 2)
````

````python
vector = np.array([1.0, 2.0, 3.0])

print("Array:", vector) # Print the array itself
print("Shape:", vector.shape) # Print the shape of the array; the number of elements along each dimension
print("Number of dimensions:", vector.ndim) # Print the number of dimensions
print("Number of elements:", vector.size) # Print the total number of elements
print("Data type:", vector.dtype) # Print the data type of the elements
````

````python

matrix = np.array([
    [1.0, 0.0, 2.0],
    [0.0, 1.0, -1.0],
])

print(matrix)
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype) # Print the data type of the elements

# Python numpy tutorial:
# https://numpy.org/doc/stable/user/quickstart.html
````

````python
# Matrix operations

matrix = np.array([
    [1.0, 0.0, 2.0],
    [0.0, 1.0, -1.0]
])

vector = np.array([1.0, 2.0, 3.0])

result = matrix @ vector

print(f"matrix-vector product: {result}")

print(f"the dimensions of the matrix: {matrix.shape} @ the dimensions of the vector: {vector.shape} --> {result.shape}")
````
